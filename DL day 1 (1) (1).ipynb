```python
dataset_encoder=dataset.copy()
```

```python
from sklearn.preprocessing import LabelEncoder

for col in dataset_encoder.select_dtypes(include="object").columns:
    dataset_encoder[col] = LabelEncoder().fit_transform(dataset_encoder[col])
```

```python
#types of neural network
#perceptron
#mlxtend is powerful python library built on top of numpy , scipy,scikitlearn
#it offer helpfull tool for machine learning tasks such as model evaluation, feature selection,assemble methods,data visualization
```

```python
x=dataset_encoder.drop("placed",axis=1)
y=dataset_encoder["placed"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from  sklearn.linear_model import Perceptron
pr = Perceptron()
pr.fit(x_train,y_train)
```

```python
pr.score(x_train,y_train)*100,pr.score(x_test,y_test)*100
```

```python
plt.figure(figsize=(4,3))
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf = pr)
plt.show()

```
