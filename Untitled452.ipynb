```python
x = "NLP research has helped enable the era of generative AI, from the communication skills of large language models (LLMs) to the ability of image generation models to understand requests."
```

```python
x
```

```python
from nltk.tokenize import word_tokenize
```

```python
from nltk.util import ngrams
```

```python
w = word_tokenize(x)
```

```python
w
```

```python
from nltk.collocations import BigramCollocationFinder,TrigramCollocationFinder
```

```python
from nltk.util import ngrams
```

```python
b = BigramCollocationFinder.from_words(w)
```

```python
b.ngram_fd
```

```python
t = TrigramCollocationFinder.from_words(w)
```

```python
t.ngram_fd
```

```python
n = ngrams(w,3)
```

```python
for i in n:
    print(i)
```
